# Stage 5 — Обучение обоих вариантов

Тренируем `FlattenMLP` (A) и `FrameRNN` (B) на одних и тех же train/val/test сплитах и сравниваем. Единый тренировочный цикл, разный тип модели.

- AdamW + cosine LR
- Weighted CE (class_weights из stage3)
- Early stopping по val macro-F1
- Отдельный best checkpoint на вариант

```python
# ============================================================
# 5.0 — Импорты, ClearML и конфиг
# ============================================================
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.nn.utils.rnn import pack_padded_sequence
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import f1_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import json, time
from pathlib import Path

from clearml import Task, Dataset as ClearMLDataset
task = Task.init(
    project_name="Video Classifier",
    task_name="05_training",
    task_type=Task.TaskTypes.training,
    reuse_last_task_id=False,
    auto_connect_frameworks={'pytorch': True, 'matplotlib': True},
)

try:
    SRC_DIR = Path(ClearMLDataset.get(
        dataset_name="yt8m-frame-processed",
        dataset_project="Video Classifier",
    ).get_local_copy())
    STAGE3_DIR = Path(ClearMLDataset.get(
        dataset_name="yt8m-frame-splits",
        dataset_project="Video Classifier",
    ).get_local_copy())
    print(f'✓ processed: {SRC_DIR}')
    print(f'✓ splits   : {STAGE3_DIR}')
except Exception as e:
    SRC_DIR    = Path('../data/frame_processed')
    STAGE3_DIR = Path('../data/frame_stage3')
    print(f'⚠ ClearML Dataset get failed ({e}); using local dirs')

STAGE4_DIR = Path('../data/frame_stage4')
STAGE5_DIR = Path('../data/frame_stage5'); STAGE5_DIR.mkdir(exist_ok=True)

cfg_src = json.load(open(SRC_DIR    / 'config.json'))
cfg     = json.load(open(STAGE3_DIR / 'config.json'))

try:
    train_cfg = json.load(open(STAGE4_DIR / 'train_config.json'))
except FileNotFoundError:
    train_cfg = {
        'variants': ['flatten', 'rnn', 'gated_rnn'], 'batch_size': 256, 'epochs': 25,
        'lr': 3e-4, 'wd': 1e-4, 'label_smooth': 0.05, 'sched': 'cosine', 'seed': 42,
    }

task.connect(train_cfg, name='train_config')
model_hparams = {
    "flatten":   {"hidden": [1024, 512], "dropout": 0.4, "n_frames_use": 30},
    "rnn":       {"proj_dim": 256, "rnn_hidden": 256, "rnn_layers": 1,
                  "bidirectional": True, "dropout": 0.3, "cell": "gru"},
    "gated_rnn": {"proj_dim": 256, "rnn_hidden": 256, "rnn_layers": 1,
                  "bidirectional": True, "dropout": 0.3, "cell": "gru"},
}
task.connect(model_hparams, name='model_hparams')

L         = cfg['L']
DIM_RGB   = cfg['dim_rgb']
DIM_AUD   = cfg['dim_audio']
N_CLASSES = cfg['n_classes']
GENRES    = cfg['genres']
DEQ_S     = cfg_src['dequant_scale']
DEQ_B     = cfg_src['dequant_bias']
DEVICE    = torch.device(
    'cuda' if torch.cuda.is_available()
    else 'mps' if torch.backends.mps.is_available()
    else 'cpu'
)

torch.manual_seed(train_cfg['seed']); np.random.seed(train_cfg['seed'])
print('device:', DEVICE, ' L:', L, ' classes:', N_CLASSES, ' task:', task.id)
```

```python
# ============================================================
# 5.1 — FrameDataset + GatedFrameDataset
# ============================================================
X_rgb_q   = np.load(SRC_DIR / 'X_rgb.npy',   mmap_mode='r')
X_audio_q = np.load(SRC_DIR / 'X_audio.npy', mmap_mode='r')
lengths   = np.load(SRC_DIR / 'lengths.npy')
y         = np.load(SRC_DIR / 'y.npy')
idx_train = np.load(STAGE3_DIR / 'idx_train.npy')
idx_val   = np.load(STAGE3_DIR / 'idx_val.npy')
idx_test  = np.load(STAGE3_DIR / 'idx_test.npy')
with np.load(STAGE3_DIR / 'norm_stats.npz') as _f:
    norm = {k: _f[k].astype(np.float32) for k in _f.files}
class_w = torch.load(STAGE3_DIR / 'class_weights.pt', weights_only=True).to(DEVICE)


class FrameDataset(Dataset):
    def __init__(self, indices, augment=False, drop_audio_p=0.0, use_audio=True):
        self.indices      = np.asarray(indices, dtype=np.int64)
        self.augment      = augment
        self.drop_audio_p = drop_audio_p
        self.use_audio    = use_audio

    def __len__(self): return len(self.indices)

    def __getitem__(self, i):
        idx = int(self.indices[i]); T = int(lengths[idx])
        rgb = np.asarray(X_rgb_q[idx],   dtype=np.float32) * DEQ_S + DEQ_B
        aud = np.asarray(X_audio_q[idx], dtype=np.float32) * DEQ_S + DEQ_B
        rgb[:T] = (rgb[:T] - norm['rgb_mean']) / norm['rgb_std']
        rgb[T:] = 0
        if self.use_audio:
            aud[:T] = (aud[:T] - norm['aud_mean']) / norm['aud_std']
            aud[T:] = 0
            if self.drop_audio_p and np.random.rand() < self.drop_audio_p:
                aud[:] = 0.0
        else:
            aud[:] = 0.0
        if self.augment:
            rgb[:T] += np.random.randn(*rgb[:T].shape).astype(np.float32) * 0.02
            if self.use_audio:
                aud[:T] += np.random.randn(*aud[:T].shape).astype(np.float32) * 0.03
        return (torch.from_numpy(rgb), torch.from_numpy(aud),
                torch.tensor(T, dtype=torch.long),
                torch.tensor(int(y[idx]), dtype=torch.long))


class GatedFrameDataset(Dataset):
    """Как FrameDataset, но audio *= gate_value и возвращает gate_scalar."""
    def __init__(self, indices, gate_values, augment=False):
        self.indices   = np.asarray(indices, dtype=np.int64)
        self.gate_vals = gate_values.astype(np.float32)
        self.augment   = augment

    def __len__(self): return len(self.indices)

    def __getitem__(self, i):
        idx = int(self.indices[i]); T = int(lengths[idx])
        g   = float(self.gate_vals[i])
        rgb = np.asarray(X_rgb_q[idx],   dtype=np.float32) * DEQ_S + DEQ_B
        aud = np.asarray(X_audio_q[idx], dtype=np.float32) * DEQ_S + DEQ_B
        rgb[:T] = (rgb[:T] - norm['rgb_mean']) / norm['rgb_std']; rgb[T:] = 0
        aud[:T] = (aud[:T] - norm['aud_mean']) / norm['aud_std']; aud[T:] = 0
        aud *= g
        if self.augment:
            rgb[:T] += np.random.randn(*rgb[:T].shape).astype(np.float32) * 0.02
            if g > 0:
                aud[:T] += np.random.randn(*aud[:T].shape).astype(np.float32) * 0.03
        return (torch.from_numpy(rgb), torch.from_numpy(aud),
                torch.tensor(g, dtype=torch.float32),
                torch.tensor(T, dtype=torch.long),
                torch.tensor(int(y[idx]), dtype=torch.long))


BATCH = train_cfg['batch_size']
# num_workers=0 на macOS: spawn не может сериализовать классы из ноутбука;
# на CUDA — 2 для асинхронной подгрузки
NUM_WORKERS = 2 if DEVICE.type == 'cuda' else 0
dl_kw = dict(batch_size=BATCH, num_workers=NUM_WORKERS, pin_memory=(DEVICE.type == 'cuda'))
train_loader = DataLoader(FrameDataset(idx_train, augment=True, drop_audio_p=0.1),
                          shuffle=True,  **dl_kw)
val_loader   = DataLoader(FrameDataset(idx_val),  shuffle=False, **dl_kw)
test_loader  = DataLoader(FrameDataset(idx_test), shuffle=False, **dl_kw)
print(f'loaders: train={len(train_loader)} val={len(val_loader)} test={len(test_loader)}  batch={BATCH}  workers={NUM_WORKERS}')
```

```python
# ============================================================
# 5.2 — Определения моделей (самодостаточно)
# ============================================================
class FlattenMLP(nn.Module):
    def __init__(self, n_frames, dim_rgb, dim_aud, n_classes,
                 hidden=(1024, 512), dropout=0.4):
        super().__init__()
        self.n_frames = n_frames
        in_dim = n_frames * (dim_rgb + dim_aud)
        layers = []; prev = in_dim
        for h in hidden:
            layers += [nn.Linear(prev, h), nn.BatchNorm1d(h),
                       nn.GELU(), nn.Dropout(dropout)]
            prev = h
        layers.append(nn.Linear(prev, n_classes))
        self.net = nn.Sequential(*layers)

    def forward(self, rgb, aud, lengths=None):
        K = self.n_frames
        x = torch.cat([rgb[:, :K], aud[:, :K]], dim=-1).reshape(rgb.size(0), -1)
        return self.net(x)


class FrameRNN(nn.Module):
    def __init__(self, dim_rgb, dim_aud, n_classes,
                 proj_dim=256, rnn_hidden=256, rnn_layers=1,
                 bidirectional=True, dropout=0.3, cell='gru'):
        super().__init__()
        self.proj = nn.Sequential(nn.Linear(dim_rgb + dim_aud, proj_dim),
                                  nn.LayerNorm(proj_dim), nn.GELU(), nn.Dropout(dropout))
        RNN = {'gru': nn.GRU, 'lstm': nn.LSTM}[cell]
        self.rnn = RNN(proj_dim, rnn_hidden, num_layers=rnn_layers,
                       batch_first=True, bidirectional=bidirectional,
                       dropout=dropout if rnn_layers > 1 else 0.0)
        out_dim = rnn_hidden * (2 if bidirectional else 1)
        self.head = nn.Sequential(nn.LayerNorm(out_dim), nn.Dropout(dropout),
                                  nn.Linear(out_dim, n_classes))
        self.bi = bidirectional; self.cell = cell

    def forward(self, rgb, aud, lengths):
        x = self.proj(torch.cat([rgb, aud], dim=-1))
        lens_cpu = lengths.detach().cpu().clamp(min=1)
        packed = pack_padded_sequence(x, lens_cpu, batch_first=True, enforce_sorted=False)
        _, h = self.rnn(packed)
        if self.cell == 'lstm': h = h[0]
        h = torch.cat([h[-2], h[-1]], dim=-1) if self.bi else h[-1]
        return self.head(h)


class AudioGateNet(nn.Module):
    def __init__(self, dim_aud, hidden=128, dropout=0.3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(dim_aud, hidden), nn.BatchNorm1d(hidden), nn.GELU(),
            nn.Dropout(dropout), nn.Linear(hidden, 64), nn.GELU(), nn.Linear(64, 1),
        )

    def forward(self, x):
        return self.net(x).squeeze(-1)

    def gate(self, audio_seq, lengths, threshold=0.5):
        T = lengths.float().clamp(min=1)
        mask = torch.arange(audio_seq.size(1), device=audio_seq.device).unsqueeze(0) < lengths.unsqueeze(1)
        mean_aud = (audio_seq * mask.unsqueeze(-1)).sum(1) / T.unsqueeze(1)
        with torch.no_grad():
            prob = torch.sigmoid(self.forward(mean_aud))
        return (prob >= threshold).float()


class GatedRNN(nn.Module):
    def __init__(self, dim_rgb, dim_aud, n_classes,
                 proj_dim=256, rnn_hidden=256, rnn_layers=1,
                 bidirectional=True, dropout=0.3, cell='gru'):
        super().__init__()
        self.proj = nn.Sequential(nn.Linear(dim_rgb + dim_aud, proj_dim),
                                  nn.LayerNorm(proj_dim), nn.GELU(), nn.Dropout(dropout))
        RNN = {'gru': nn.GRU, 'lstm': nn.LSTM}[cell]
        self.rnn = RNN(proj_dim, rnn_hidden, num_layers=rnn_layers,
                       batch_first=True, bidirectional=bidirectional,
                       dropout=dropout if rnn_layers > 1 else 0.0)
        out_dim = rnn_hidden * (2 if bidirectional else 1) + 1
        self.head = nn.Sequential(nn.LayerNorm(out_dim), nn.Dropout(dropout),
                                  nn.Linear(out_dim, n_classes))
        self.bi = bidirectional; self.cell = cell

    def forward(self, rgb, aud, gate_s, lengths):
        x = self.proj(torch.cat([rgb, aud], dim=-1))
        lens_cpu = lengths.detach().cpu().clamp(min=1)
        packed = pack_padded_sequence(x, lens_cpu, batch_first=True, enforce_sorted=False)
        _, h = self.rnn(packed)
        if self.cell == 'lstm': h = h[0]
        h = torch.cat([h[-2], h[-1]], dim=-1) if self.bi else h[-1]
        h = torch.cat([h, gate_s.unsqueeze(1)], dim=-1)
        return self.head(h)


class CrossModalRNN(nn.Module):
    """RGB кадры запрашивают аудио через cross-attention (Flamingo-style).
    Modality dropout обучает модель работать и без аудио."""
    def __init__(self, dim_rgb, dim_aud, n_classes,
                 proj_dim=256, num_heads=4,
                 rnn_hidden=256, rnn_layers=1, dropout=0.3):
        super().__init__()
        self.proj_rgb = nn.Sequential(
            nn.Linear(dim_rgb, proj_dim), nn.LayerNorm(proj_dim), nn.GELU(),
        )
        self.proj_aud = nn.Sequential(
            nn.Linear(dim_aud, proj_dim), nn.LayerNorm(proj_dim), nn.GELU(),
        )
        self.cross_attn = nn.MultiheadAttention(
            embed_dim=proj_dim, num_heads=num_heads,
            dropout=dropout, batch_first=True,
        )
        self.attn_norm = nn.LayerNorm(proj_dim)
        self.attn_drop = nn.Dropout(dropout)
        self.rnn = nn.GRU(
            proj_dim, rnn_hidden, num_layers=rnn_layers,
            batch_first=True, bidirectional=True,
            dropout=dropout if rnn_layers > 1 else 0.0,
        )
        self.head = nn.Sequential(
            nn.LayerNorm(rnn_hidden * 2), nn.Dropout(dropout),
            nn.Linear(rnn_hidden * 2, n_classes),
        )

    def forward(self, rgb, aud, lengths, mod_dropout_p=0.0):
        B, L, _ = rgb.shape
        if self.training and mod_dropout_p > 0:
            drop_mask = torch.rand(B, 1, 1, device=rgb.device) < mod_dropout_p
            aud = aud.masked_fill(drop_mask, 0.0)
        q = self.proj_rgb(rgb)
        k = self.proj_aud(aud)
        # lengths приходит на CPU из DataLoader — переносим на устройство модели
        pad_mask = torch.arange(L, device=rgb.device).unsqueeze(0) >= lengths.to(rgb.device).unsqueeze(1)
        attended, _ = self.cross_attn(q, k, k, key_padding_mask=pad_mask)
        x = self.attn_norm(q + self.attn_drop(attended))
        lens_cpu = lengths.detach().cpu().clamp(min=1)
        packed = pack_padded_sequence(x, lens_cpu, batch_first=True, enforce_sorted=False)
        _, h = self.rnn(packed)
        h = torch.cat([h[-2], h[-1]], dim=-1)
        return self.head(h)


def build(kind):
    if kind == 'flatten':
        return FlattenMLP(min(30, L), DIM_RGB, DIM_AUD, N_CLASSES,
                          hidden=(1024, 512), dropout=0.4)
    if kind == 'rnn':
        return FrameRNN(DIM_RGB, DIM_AUD, N_CLASSES,
                        proj_dim=256, rnn_hidden=256, bidirectional=True,
                        dropout=0.3, cell='gru')
    if kind == 'gated_rnn':
        return GatedRNN(DIM_RGB, DIM_AUD, N_CLASSES,
                        proj_dim=256, rnn_hidden=256, bidirectional=True,
                        dropout=0.3, cell='gru')
    if kind == 'cross_modal':
        return CrossModalRNN(DIM_RGB, DIM_AUD, N_CLASSES,
                             proj_dim=256, num_heads=4, rnn_hidden=256,
                             rnn_layers=1, dropout=0.3)
    raise ValueError(kind)
```

```python
# ============================================================
# 5.2b — Gate pre-training + 5.3 — Train / Eval функции
# ============================================================
from torch.utils.data import TensorDataset as TDS

class TinyRNN(nn.Module):
    def __init__(self, dim_rgb, dim_aud, n_classes,
                 proj_dim=64, rnn_hidden=64, dropout=0.2):
        super().__init__()
        self.proj = nn.Sequential(nn.Linear(dim_rgb + dim_aud, proj_dim),
                                  nn.LayerNorm(proj_dim), nn.GELU())
        self.rnn  = nn.GRU(proj_dim, rnn_hidden, batch_first=True, bidirectional=True)
        self.head = nn.Sequential(nn.LayerNorm(rnn_hidden * 2), nn.Dropout(dropout),
                                  nn.Linear(rnn_hidden * 2, n_classes))

    def forward(self, rgb, aud, lengths):
        x = self.proj(torch.cat([rgb, aud], dim=-1))
        lens_cpu = lengths.detach().cpu().clamp(min=1)
        _, h = self.rnn(pack_padded_sequence(x, lens_cpu, batch_first=True, enforce_sorted=False))
        return self.head(torch.cat([h[-2], h[-1]], dim=-1))


def _quick_train(model, tr_loader, epochs=10, lr=1e-3):
    opt = torch.optim.Adam(model.parameters(), lr=lr)
    crit = nn.CrossEntropyLoss()
    for _ in range(epochs):
        model.train()
        for rgb, aud, lens, yt in tr_loader:
            rgb, aud, yt = rgb.to(DEVICE), aud.to(DEVICE), yt.to(DEVICE)
            opt.zero_grad(); crit(model(rgb, aud, lens), yt).backward(); opt.step()
    return model


@torch.no_grad()
def _get_preds(model, indices, use_audio):
    model.eval()
    preds, trues = [], []
    dl = DataLoader(FrameDataset(indices, use_audio=use_audio),
                    batch_size=512, shuffle=False, num_workers=NUM_WORKERS)
    for rgb, aud, lens, yt in dl:
        p = model(rgb.to(DEVICE), aud.to(DEVICE), lens).argmax(1).cpu().numpy()
        preds.extend(p); trues.extend(yt.numpy())
    return np.array(preds), np.array(trues)


def _audio_mean_pool(indices):
    feats = []
    for idx in indices:
        T   = int(lengths[idx])
        aud = np.asarray(X_audio_q[idx, :max(T, 1)], dtype=np.float32) * DEQ_S + DEQ_B
        aud = (aud - norm['aud_mean']) / norm['aud_std']
        feats.append(aud.mean(0))
    return np.stack(feats)


print('Gate pre-training: TinyRNN × 2 ...')
GATE_EPOCHS = 10
tiny_no  = _quick_train(TinyRNN(DIM_RGB, DIM_AUD, N_CLASSES).to(DEVICE),
                        DataLoader(FrameDataset(idx_train, use_audio=False),
                                   batch_size=BATCH, shuffle=True, num_workers=NUM_WORKERS), GATE_EPOCHS)
tiny_yes = _quick_train(TinyRNN(DIM_RGB, DIM_AUD, N_CLASSES).to(DEVICE),
                        DataLoader(FrameDataset(idx_train, use_audio=True),
                                   batch_size=BATCH, shuffle=True, num_workers=NUM_WORKERS), GATE_EPOCHS)

p_no,  true_tr = _get_preds(tiny_no,  idx_train, use_audio=False)
p_yes, _       = _get_preds(tiny_yes, idx_train, use_audio=True)
audio_useful   = ((p_yes == true_tr) & (p_no != true_tr)).astype(np.int64)
print(f'  audio helpful on train: {audio_useful.mean():.3f}  ({audio_useful.sum()}/{len(audio_useful)})')

X_ga_tr = _audio_mean_pool(idx_train)
X_ga_va = _audio_mean_pool(idx_val)
X_ga_te = _audio_mean_pool(idx_test)

gate_net = AudioGateNet(DIM_AUD).to(DEVICE)
pos_w    = torch.tensor(
    [(audio_useful == 0).sum() / max((audio_useful == 1).sum(), 1)],
    dtype=torch.float32).to(DEVICE)
g_opt  = torch.optim.Adam(gate_net.parameters(), lr=1e-3, weight_decay=1e-4)
g_crit = nn.BCEWithLogitsLoss(pos_weight=pos_w)
g_dl   = DataLoader(TDS(torch.tensor(X_ga_tr, dtype=torch.float32),
                        torch.tensor(audio_useful, dtype=torch.float32)),
                    batch_size=512, shuffle=True)
gate_net.train()
for _ in range(20):
    for Xb, yb in g_dl:
        Xb, yb = Xb.to(DEVICE), yb.to(DEVICE)
        g_opt.zero_grad(); g_crit(gate_net(Xb), yb).backward(); g_opt.step()

torch.save(gate_net.state_dict(), STAGE5_DIR / 'gate_net.pt')
print('Gate net saved.')


def _compute_gates(X_gate, threshold=0.5):
    gate_net.eval()
    with torch.no_grad():
        prob = torch.sigmoid(gate_net(torch.tensor(X_gate, dtype=torch.float32).to(DEVICE)))
    return (prob.cpu().numpy() >= threshold).astype(np.float32)

gates_train = _compute_gates(X_ga_tr)
gates_val   = _compute_gates(X_ga_va)
gates_test  = _compute_gates(X_ga_te)
print(f'Gate rates — train:{gates_train.mean():.3f}  val:{gates_val.mean():.3f}  test:{gates_test.mean():.3f}')

_kw = dict(num_workers=NUM_WORKERS, pin_memory=(DEVICE.type == 'cuda'))
gated_train_loader = DataLoader(GatedFrameDataset(idx_train, gates_train, augment=True),
                                batch_size=BATCH, shuffle=True,  **_kw)
gated_val_loader   = DataLoader(GatedFrameDataset(idx_val,   gates_val),
                                batch_size=BATCH, shuffle=False, **_kw)
gated_test_loader  = DataLoader(GatedFrameDataset(idx_test,  gates_test),
                                batch_size=BATCH, shuffle=False, **_kw)
print('Gated loaders ready.')


# ── Train / Eval ─────────────────────────────────────────────

@torch.no_grad()
def evaluate(model, loader, is_gated=False):
    model.eval()
    total_loss = n = correct = 0
    preds, trues = [], []
    for batch in loader:
        if is_gated:
            rgb, aud, gate_s, lens, yt = batch
            gate_s = gate_s.to(DEVICE, non_blocking=True)
        else:
            rgb, aud, lens, yt = batch
        rgb = rgb.to(DEVICE, non_blocking=True)
        aud = aud.to(DEVICE, non_blocking=True)
        yt  = yt.to(DEVICE,  non_blocking=True)
        logits = model(rgb, aud, gate_s, lens) if is_gated else model(rgb, aud, lens)
        loss = F.cross_entropy(logits, yt, weight=class_w,
                               label_smoothing=train_cfg['label_smooth'])
        pred = logits.argmax(-1)
        total_loss += loss.item() * yt.size(0)
        correct    += (pred == yt).sum().item()
        n          += yt.size(0)
        preds.append(pred.cpu().numpy()); trues.append(yt.cpu().numpy())
    preds = np.concatenate(preds); trues = np.concatenate(trues)
    f1 = f1_score(trues, preds, average='macro', zero_division=0)
    return total_loss / n, correct / n, f1, preds, trues


def train_one(kind, epochs=None, patience=8):
    epochs = epochs or train_cfg['epochs']
    model  = build(kind).to(DEVICE)
    nparams = sum(p.numel() for p in model.parameters())
    print(f'\n=== Training [{kind}] — {nparams/1e6:.2f}M params ===')

    is_gated      = (kind == 'gated_rnn')
    MOD_DROPOUT_P = 0.3 if kind == 'cross_modal' else 0.0
    tr_loader_use = gated_train_loader if is_gated else train_loader
    va_loader_use = gated_val_loader   if is_gated else val_loader

    logger = task.get_logger()
    logger.report_single_value(f'{kind}/n_params_M', round(nparams / 1e6, 3))

    opt   = torch.optim.AdamW(model.parameters(), lr=train_cfg['lr'],
                               weight_decay=train_cfg['wd'])
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=epochs)
    hist  = {'train_loss': [], 'val_loss': [], 'train_acc': [], 'val_acc': [], 'val_f1': []}
    best_f1 = -1; stale = 0

    for ep in range(1, epochs + 1):
        model.train()
        t0 = time.time(); t_loss = t_correct = t_n = 0
        for batch in tr_loader_use:
            if is_gated:
                rgb, aud, gate_s, lens, yt = batch
                gate_s = gate_s.to(DEVICE, non_blocking=True)
            else:
                rgb, aud, lens, yt = batch
            rgb = rgb.to(DEVICE, non_blocking=True)
            aud = aud.to(DEVICE, non_blocking=True)
            yt  = yt.to(DEVICE,  non_blocking=True)
            if is_gated:
                logits = model(rgb, aud, gate_s, lens)
            elif MOD_DROPOUT_P > 0:
                logits = model(rgb, aud, lens, mod_dropout_p=MOD_DROPOUT_P)
            else:
                logits = model(rgb, aud, lens)
            loss = F.cross_entropy(logits, yt, weight=class_w,
                                   label_smoothing=train_cfg['label_smooth'])
            opt.zero_grad(); loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            opt.step()
            t_loss    += loss.item() * yt.size(0)
            t_correct += (logits.argmax(-1) == yt).sum().item()
            t_n       += yt.size(0)
        sched.step()
        v_loss, v_acc, v_f1, *_ = evaluate(model, va_loader_use, is_gated=is_gated)

        tr_loss = t_loss / t_n; tr_acc = t_correct / t_n
        hist['train_loss'].append(tr_loss); hist['train_acc'].append(tr_acc)
        hist['val_loss'].append(v_loss);    hist['val_acc'].append(v_acc)
        hist['val_f1'].append(v_f1)

        logger.report_scalar('loss',     f'{kind}/train', value=tr_loss, iteration=ep)
        logger.report_scalar('loss',     f'{kind}/val',   value=v_loss,  iteration=ep)
        logger.report_scalar('accuracy', f'{kind}/train', value=tr_acc,  iteration=ep)
        logger.report_scalar('accuracy', f'{kind}/val',   value=v_acc,   iteration=ep)
        logger.report_scalar('macro_f1', f'{kind}/val',   value=v_f1,    iteration=ep)
        logger.report_scalar('lr',       kind,            value=opt.param_groups[0]['lr'], iteration=ep)

        print(f'  ep{ep:>3}  train_loss={tr_loss:.4f} acc={tr_acc:.3f} | '
              f'val_loss={v_loss:.4f} acc={v_acc:.3f} f1={v_f1:.3f}  ({time.time()-t0:.1f}s)')
        if v_f1 > best_f1:
            best_f1 = v_f1; stale = 0
            torch.save(model.state_dict(), STAGE5_DIR / f'best_{kind}.pt')
        else:
            stale += 1
            if stale >= patience:
                print(f'  early stop @ ep{ep} (best val_f1={best_f1:.3f})')
                break

    logger.report_single_value(f'{kind}/best_val_macro_f1', round(best_f1, 4))
    return model, hist, best_f1
```

```python
# ============================================================
# 5.4 — Прогон всех вариантов
# ============================================================
results = {}
logger = task.get_logger()
for kind in train_cfg['variants']:
    torch.manual_seed(train_cfg['seed']); np.random.seed(train_cfg['seed'])
    model, hist, best = train_one(kind)
    model.load_state_dict(torch.load(STAGE5_DIR / f'best_{kind}.pt', weights_only=True))
    is_gated  = (kind == 'gated_rnn')
    te_loader = gated_test_loader if is_gated else test_loader
    t_loss, t_acc, t_f1, preds, trues = evaluate(model, te_loader, is_gated=is_gated)
    results[kind] = dict(history=hist, best_val_f1=best,
                         test_loss=t_loss, test_acc=t_acc, test_f1=t_f1,
                         preds=preds.tolist(), trues=trues.tolist())
    logger.report_single_value(f'{kind}/test_loss',     round(t_loss, 4))
    logger.report_single_value(f'{kind}/test_accuracy', round(t_acc,  4))
    logger.report_single_value(f'{kind}/test_macro_f1', round(t_f1,   4))
    print(f'[{kind}]  test: loss={t_loss:.4f}  acc={t_acc:.3f}  macro-f1={t_f1:.3f}')
```

```python
# ============================================================
# 5.5 — Per-class отчёт и confusion matrix для лучшего варианта
# ============================================================
best_kind = max(train_cfg['variants'], key=lambda k: results[k]['test_f1'])
print(f'best variant on test: {best_kind}  (macro-f1 = {results[best_kind]["test_f1"]:.3f})\n')

preds = np.array(results[best_kind]['preds'])
trues = np.array(results[best_kind]['trues'])
present = np.unique(trues)
print(classification_report(trues, preds, labels=present,
                             target_names=[GENRES[i] for i in present],
                             zero_division=0))

cm = confusion_matrix(trues, preds, labels=np.arange(N_CLASSES), normalize='true')
fig, ax = plt.subplots(figsize=(8, 7))
im = ax.imshow(cm, cmap='Blues', vmin=0, vmax=1)
ax.set_xticks(range(N_CLASSES)); ax.set_yticks(range(N_CLASSES))
ax.set_xticklabels(GENRES, rotation=45, ha='right', fontsize=8)
ax.set_yticklabels(GENRES, fontsize=8)
ax.set_title(f'Confusion Matrix ({best_kind}, normalized by true)')
for i in range(N_CLASSES):
    for j in range(N_CLASSES):
        v = cm[i, j]
        if v > 0.02:
            ax.text(j, i, f'{v:.2f}', ha='center', va='center',
                    color='white' if v > 0.5 else 'black', fontsize=7)
plt.colorbar(im, ax=ax, fraction=0.046)
plt.tight_layout()
plt.savefig(STAGE5_DIR / 'test_confusion_matrix.png', bbox_inches='tight', facecolor='#0F1117')
plt.show()

with open(STAGE5_DIR / 'results.json', 'w') as f:
    json.dump({k: {kk: vv for kk, vv in v.items() if kk != 'history'}
               | {'history': v['history']}
               for k, v in results.items()}, f, indent=2, default=float)
print('saved to', STAGE5_DIR)
```
